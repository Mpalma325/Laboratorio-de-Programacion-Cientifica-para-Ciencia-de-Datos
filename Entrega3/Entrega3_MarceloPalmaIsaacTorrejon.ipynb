### 📌 Análisis individual [3.0 puntos]

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import drive
drive.mount('/content/drive')

BASE_PATH = "/content/drive/MyDrive/Entrega3"
````

````python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from collections import Counter
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

clientes_df  = pd.read_parquet(f"{BASE_PATH}/clientes.parquet")
productos_df = pd.read_parquet(f"{BASE_PATH}/productos.parquet")
universe_size = len(clientes_df) * len(productos_df)

weeks = {
    53: {'pred': f'{BASE_PATH}/preds_week_53.parquet', 'gt': f'{BASE_PATH}/batch_t1.parquet'},
    54: {'pred': f'{BASE_PATH}/preds_week_54.parquet', 'gt': f'{BASE_PATH}/batch_t2.parquet'},
    55: {'pred': f'{BASE_PATH}/preds_week_55.parquet', 'gt': f'{BASE_PATH}/batch_t3.parquet'},
    56: {'pred': f'{BASE_PATH}/preds_week_56.parquet', 'gt': f'{BASE_PATH}/batch_t4.parquet'},
}

def get_metrics(pred_path, gt_path, universe_size):
    preds = pd.read_parquet(pred_path)
    pred_pairs = set(
        map(tuple, preds[preds['pred_compra'] == 1][['customer_id', 'product_id']].values)
    )

    gt = pd.read_parquet(gt_path)
    gt = gt.groupby(['customer_id', 'product_id']).size().reset_index()
    true_pairs = set(map(tuple, gt[['customer_id', 'product_id']].values))

    # TP, FP, FN
    tp = len(pred_pairs & true_pairs)
    fp = len(pred_pairs - true_pairs)
    fn = len(true_pairs - pred_pairs)
    tn = universe_size - tp - fp - fn

    y_true = [1]*tp + [0]*fp + [1]*fn + [0]*tn
    y_pred = [1]*tp + [1]*fp + [0]*fn + [0]*tn

    accuracy  = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred)
    recall    = recall_score(y_true, y_pred)
    f1        = f1_score(y_true, y_pred)

    return {
        'week': 0,
        'tp': tp,
        'fp': fp,
        'fn': fn,
        'tn': tn,
        'accuracy': accuracy,
        'precision': precision,
        'recall': recall,
        'f1': f1,
        'n_pred': len(pred_pairs),
        'n_real': len(true_pairs),
        'universe_size': universe_size
    }

def analyze_precision_distribution(pred_df, gt_df):
    """Analiza la distribución de precisión por cliente"""
    merged = pred_df.merge(gt_df.assign(label=1),
                           on=['customer_id', 'product_id'],
                           how='left')
    merged['label'] = merged['label'].fillna(0)

    per_client = merged.groupby('customer_id')['label'].agg(['sum', 'count'])
    per_client['precision'] = per_client['sum'] / per_client['count']

    # Calcular porcentajes
    pct_zero = (per_client['precision'] == 0).sum() / len(per_client) * 100
    pct_above_005 = (per_client['precision'] > 0.05).sum() / len(per_client) * 100
    pct_above_01 = (per_client['precision'] > 0.1).sum() / len(per_client) * 100

    return {
        'pct_zero_precision': pct_zero,
        'pct_above_005': pct_above_005,
        'pct_above_01': pct_above_01,
        'mean_precision': per_client['precision'].mean(),
        'per_client_stats': per_client
    }

def top_n_analysis(pred_df, gt_df, n_values=[100, 500, 1000, 5000, 10000]):
    """Analiza métricas limitando a las top-N predicciones con mayor probabilidad"""
    results = []

    for n in n_values:
        # Tomar top-N predicciones por score
        top_n_pred = pred_df.nlargest(n, 'score')[['customer_id', 'product_id']]

        # Calcular métricas
        pred_pairs = set(map(tuple, top_n_pred.values))
        true_pairs = set(map(tuple, gt_df[['customer_id', 'product_id']].values))

        tp = len(pred_pairs & true_pairs)
        fp = len(pred_pairs - true_pairs)
        fn = len(true_pairs - pred_pairs)

        precision = tp / (tp + fp) if (tp + fp) > 0 else 0
        recall = tp / (tp + fn) if (tp + fn) > 0 else 0
        f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

        results.append({
            'n': n,
            'precision': precision,
            'recall': recall,
            'f1': f1,
            'tp': tp,
            'fp': fp,
            'fn': fn
        })

    return pd.DataFrame(results)

def sample_clients_analysis(pred_df, gt_df, sample_size=50, seed=42):
    """Analiza una muestra de clientes comparando recomendaciones vs compras reales"""

    # Tomar muestra aleatoria de clientes que tienen predicciones
    np.random.seed(seed)
    sample_clients = np.random.choice(
        pred_df['customer_id'].unique(),
        size=min(sample_size, len(pred_df['customer_id'].unique())),
        replace=False
    )

    # Contar recomendaciones por cliente (pred_compra == 1)
    recs_per_client = pred_df[
        (pred_df['customer_id'].isin(sample_clients)) &
        (pred_df['pred_compra'] == 1)
    ].groupby('customer_id').size()

    # Contar compras reales por cliente
    purchases_per_client = gt_df[
        gt_df['customer_id'].isin(sample_clients)
    ].groupby('customer_id')['product_id'].nunique()

    # Asegurar que todos los clientes estén presentes
    all_clients = pd.DataFrame({'customer_id': sample_clients})
    all_clients = all_clients.merge(
        recs_per_client.rename('n_recomendaciones'),
        left_on='customer_id',
        right_index=True,
        how='left'
    ).fillna(0)
    all_clients = all_clients.merge(
        purchases_per_client.rename('n_compras'),
        left_on='customer_id',
        right_index=True,
        how='left'
    ).fillna(0)

    stats = {
        'sample_size': len(sample_clients),
        'recs_min': int(all_clients['n_recomendaciones'].min()),
        'recs_max': int(all_clients['n_recomendaciones'].max()),
        'recs_mean': all_clients['n_recomendaciones'].mean(),
        'purchases_min': int(all_clients['n_compras'].min()),
        'purchases_max': int(all_clients['n_compras'].max()),
        'purchases_mean': all_clients['n_compras'].mean(),
        'ratio_mean': (all_clients['n_recomendaciones'] / all_clients['n_compras'].replace(0, 1)).mean(),
        'detail': all_clients
    }

    return stats

def batch_report(week_num, pred_df, gt_df, metrics, prev_metrics=None):
    """ Reporte por batch"""

    prec_dist = analyze_precision_distribution(pred_df, gt_df)
    top_n_df = top_n_analysis(pred_df, gt_df)

    # Análisis de muestra de clientes
    sample_stats = sample_clients_analysis(pred_df, gt_df, sample_size=50)

    report = f"""
{'='*80}
BATCH {week_num - 52} ({['01/01/25 - 05/01/25', '06/01/25 - 12/01/25', '13/01/25 - 19/01/25', '20/01/25 - 26/01/25'][week_num - 53]})
{'='*80}

Tasa de predicciones: {metrics['n_pred']/metrics['n_real']:.1f}x (emite {metrics['n_pred']/metrics['n_real']:.1f} recomendaciones por cada compra real)

Análisis por cliente:
  - {prec_dist['pct_zero_precision']:.1f}% de clientes tiene precisión igual a cero
  - {prec_dist['pct_above_005']:.1f}% de clientes supera precisión de 0.05
  - {prec_dist['pct_above_01']:.1f}% de clientes supera precisión de 0.1
  - Precisión promedio por cliente: {prec_dist['mean_precision']:.4f}

Análisis muestra de clientes (n={sample_stats['sample_size']}):

- Min de recomendaciones para un cliente: {sample_stats['recs_min']}
- Max de recomendaciones para un cliente: {sample_stats['recs_max']}
- Promedio de recomendaciones por cliente: {sample_stats['recs_mean']:.1f}
- Min de compras para un cliente: {sample_stats['purchases_min']}
- Max de compras para un cliente: {sample_stats['purchases_max']}
- Promedio de compras por cliente: {sample_stats['purchases_mean']:.1f}
- Recomendaciones por cada compra real: {sample_stats['ratio_mean']:.1f}

Análisis TOP-N (limitando a las N predicciones con mayor score):
"""

    for _, row in top_n_df.iterrows():
        report += f"  - Top {row['n']:,}: Precision={row['precision']:.4f}, Recall={row['recall']:.4f}, F1={row['f1']:.4f}\n"

    # Comparación con batch anterior
    if prev_metrics:
        report += f"""
Comparación batch anterior:
  - TP: {metrics['tp']:,} vs {prev_metrics['tp']:,} ({((metrics['tp']-prev_metrics['tp'])/prev_metrics['tp']*100):+.1f}%)
  - FP: {metrics['fp']:,} vs {prev_metrics['fp']:,} ({((metrics['fp']-prev_metrics['fp'])/prev_metrics['fp']*100):+.1f}%)
  - FN: {metrics['fn']:,} vs {prev_metrics['fn']:,} ({((metrics['fn']-prev_metrics['fn'])/prev_metrics['fn']*100):+.1f}%)
  - Recall: {metrics['recall']:.4f} vs {prev_metrics['recall']:.4f} ({(metrics['recall']-prev_metrics['recall'])*100:+.2f} puntos)
  - Precision: {metrics['precision']:.4f} vs {prev_metrics['precision']:.4f} ({(metrics['precision']-prev_metrics['precision'])*100:+.2f} puntos)
"""

    report += "\n" + "="*80 + "\n"

    return report

# Calcular métricas generales
results = []
for week, paths in weeks.items():
    metrics = get_metrics(paths['pred'], paths['gt'], universe_size)
    metrics['week'] = week
    results.append(metrics)

df = pd.DataFrame(results)
print("\n" + "="*80)
print("RESUMEN DE MÉTRICAS")
print("="*80)
print(df)

print(f"\nPromedio F1: {df['f1'].mean():.3f}")
print(f"Promedio Precision: {df['precision'].mean():.3f}")
print(f"Promedio Recall: {df['recall'].mean():.3f}")
````
