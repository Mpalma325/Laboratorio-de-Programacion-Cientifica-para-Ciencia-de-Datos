En el competitivo universo de las bebidas gaseosas, la empresa **SodAI Drinks 🥤** ha logrado destacarse por su creatividad, diversidad de productos y enfoque centrado en el cliente. Ofrece una extensa gama de bebidas carbonatadas que abarca distintos segmentos del mercado: desde productos premium en presentaciones sofisticadas, hasta gaseosas accesibles para el consumo masivo, disponibles en diversos tamaños y tipos de envases.

La compañía opera en múltiples regiones y zonas, sirviendo a una variedad de puntos de venta que incluyen desde tiendas de conveniencia y minimarkets hasta el canal frío tradicional. Cada tipo de cliente tiene sus particularidades: algunos reciben entregas hasta 4 veces por semana, mientras que otros son visitados por la fuerza de ventas solo una vez semanalmente. Esta diversidad de perfiles representa tanto una oportunidad como un desafío comercial: ¿cómo saber qué productos tienen más chances de ser comprados por cada cliente en un momento dado?

Con el objetivo de aumentar la facturación de forma inteligente y mejorar la eficiencia de su estrategia de ventas, **SodAI Drinks** decide crear una nueva célula interna de innovación: el equipo **Deep Drinkers 🤖**, cuyo propósito es aplicar ciencia de datos para anticiparse a las necesidades del cliente y potenciar el negocio desde una perspectiva basada en información.

El corazón de esta iniciativa es el desarrollo de un sistema predictivo personalizado para cada cliente. Para ello, **Deep Drinkers** convoca a un equipo de Data Scientists y especialistas en *machine learning* con una misión clara: construir un modelo predictivo que, cada semana, pueda estimar la probabilidad de compra de cada producto del portafolio para cada cliente activo.

El modelo deberá tener en cuenta múltiples factores, incluyendo:
- **Tipo de cliente**, ej. "TIENDA DE CONVENIENCIA", "MINIMARKET".
- **Frecuencia de entregas y visitas**, indicadores del nivel de actividad comercial.
- **Ubicación geográfica** (por región y zona).
- **Preferencias históricas de consumo**, inferidas por patrones de compra anteriores.
- **Características del producto**, como marca, categoría, segmento, tipo de envase y tamaño

El objetivo final es que, **cada semana**, se genere una tabla de productos priorizados: para cada cliente, un listado de productos ordenado por su probabilidad estimada de compra. Esta información será enviada al equipo comercial, que podrá usarla en call center, para incrementar las chances de concretar ventas al ofrecer justo lo que el cliente probablemente quiere comprar.

Este proyecto representa un cambio de paradigma en la forma en que **SodAI Drinks** gestiona su fuerza de ventas: de un enfoque reactivo y generalista, a uno proactivo, basado en datos y profundamente personalizado. Así, la empresa no solo espera aumentar su rentabilidad, sino también construir relaciones más sólidas con sus clientes, ofreciéndoles recomendaciones más relevantes y oportunas.

Para lograr lo anterior, el equipo **Deep Drinkers** contará con los siguientes conjuntos de datos, junto a sus respectivos atributos:

- **Datos transaccionales** (`transacciones.parquet`): contiene el historial de compras realizadas por los clientes.
	- `customer_id`: identificador único del cliente que realizó la compra.
	- `product_id`: identificador único del producto comprado.
	- `purchase_date`: fecha en que se realizó la transacción.
	- `order_id`: identificar de la orden de su pedido.
	- `payment`	monto total pagado por la transacción.

- **Datos de clientes** (`clientes.parquet`): incluye las características de cada cliente.
	- `customer_id`: identificador único del cliente.
	- `region_id`: identificador de la región geográfica donde se encuentra el cliente.
	- `customer_type`: tipo de cliente según el canal comercial, por ejemplo, “TIENDA DE CONVENIENCIA”.
	- `Y`: coordenada geográfica de latitud.
	- `X`: coordenada geográfica de longitud.
	- `num_deliver_per_week`: cantidad de entregas semanales que recibe el cliente.
	- `num_visit_per_week`: frecuencia de visitas de la fuerza de ventas por semana.

- **Datos de productos** (`productos.parquet`): describe las características de los productos del portafolio.
	- `product_id`: identificador único del producto.
	- `brand`: marca comercial del producto.
	- `category`: categoría general del producto, como “BEBIDAS CARBONATADAS”.
	- `sub_category`: subcategoría dentro de la categoría principal, por ejemplo, “GASEOSAS”.
	- `segment`: segmento de mercado al que pertenece el producto, como “PREMIUM”.
	- `package`: tipo de envase del producto.
	- `size`: tamaño del producto en litros.

Se observa que hay 885 filas duplicadas en transacciones, por lo que se eliminan a continuación.

Se observa la existencia de muchos parámetros del tipo object, por lo que por estandarización, se prefieren pasar a tipo string.

```python
clientes["customer_type"] = clientes["customer_type"].astype("string")
for col in ["brand", "category", "sub_category", "segment", "package"]:
    productos[col] = productos[col].astype("string")
```

Finalmente, al observar la descripción de las columnas, la mayoría podrían tener alguna relevancia, la cual puede ser estudiada en el eda, sin embargo es importante revisar el caso de las variables geográficas, donde hay 4 variables que pueden ser importantes.
```python
print(clientes[["region_id", "zone_id", "X", "Y"]].nunique())
```

Como se observa en los valores únicos de region_id, zone_id, X y Y, todos los registros de clientes corresponden a una sola región y zona geográfica. Al no aportar variabilidad, estas columnas sion eliminadas del conjunto de datos.

```python
clientes.drop(["region_id", "zone_id"], axis=1, inplace=True)
```

Finalmente, para poder hacer una tabla con todos los datos para poder trabajar con Machine learning, hacemos un doble merge usando como base la tabla de transacciones.

```python
data = transacciones.merge(clientes, on="customer_id", how="left")
data = data.merge(productos, on="product_id", how="left")
```

Se puede estandarizar el texto del dataset, para que no haya problemas en la  escritura de las categorías en el set de datos.

```python
for c in ["customer_type","brand","category","sub_category","segment","package"]:
    data[c] = data[c].astype("string").str.strip().str.lower().str.replace(r"\s+", " ", regex=True)
```

Finalmente, al observar los dtype de items es float, pero esto no calza con la definición de cantidad de bultos vendido , al no poder existir decimales en este.

Por lo mismo se transforma a int, haciendo aproximación al natural más cercano.

Se mantiene la versión float por si es requerida posteriormente.


```python
data["items_int"] = np.floor(data["items"] + 0.5).astype("Int64")
```

```python
# Transacciones por día.
for mes in range(1, 13):
    df_mes = data[data["purchase_date"].dt.month == mes]

    frecuencia_diaria = df_mes.groupby(df_mes["purchase_date"].dt.date).size()

    plt.figure(figsize=(12,6))
    frecuencia_diaria.plot(kind="bar", color="skyblue", edgecolor="black")
    plt.title(f"Frecuencia de transacciones por día - Mes {mes}")
    plt.xlabel("Día")
    plt.ylabel("Cantidad de transacciones")
    plt.xticks(rotation=45)
    plt.grid(axis="y", alpha=0.4)
    plt.show()
```

En este caso, se observa como la frecuencia de los registros es diaria, al haber entradas de datos todos los días del año. Al hacer los gráficos, se puede observar que en general todos los meses tienen un comportamiento similar, con meses con más ventas en general (Por ejemplo, Diciembre).
Además, se observan ciertos peaks de compra en cada mes.

```python
#Transacciones por día.
diario = data.groupby(data["purchase_date"].dt.normalize()).size()
prom = diario.groupby(diario.index.dayofweek).mean().reindex(range(7))
dias = ["Lunes","Martes","Miércoles","Jueves","Viernes","Sábado","Domingo"]

plt.figure(figsize=(9,4))
bars = plt.bar(dias, prom.values, edgecolor="black")
plt.title("Promedio de transacciones por día de la semana")
plt.xlabel("Día"); plt.ylabel("Promedio de transacciones por día")
plt.grid(axis="y", alpha=0.3)


for b, v in zip(bars, prom.values):
    plt.text(b.get_x()+b.get_width()/2, b.get_height()*1.01, f"{v:.1f}",
             ha="center", va="bottom", fontweight="bold", fontsize=6.5)

plt.tight_layout()
plt.show()
```

Si agrupamos las ventas por día, se observa una estacionalidad semanal muy marcada. Donde en general,los fines de semana la cantidad de transacciones disminuye fuertemente.

En esta sección, se deben construir pipelines para automatizar el preprocesamiento de los datos, lo cual garantizará que el flujo de trabajo sea reproducible y eficiente para esta entrega y las futuras. El objetivo es aplicar una serie de transformaciones en un orden definido para asegurar que los datos estén listos para los modelos a entrenar. El pipeline final debe incluir las técnicas de pre-procesamiento que se deben aplicar a los distintos datos (según lo que consideren necesario para el problema). Por ejemplo:

- **Imputación de valores nulos**: Manejo de datos faltantes mediante estrategias adecuadas (media, mediana, moda, interpolación, etc.).

- **Transformaciones personalizadas**: Uso de ColumnTransformer para aplicar diferentes transformaciones a columnas específicas.

- **Codificación de variables categóricas**: Convertir datos categóricos a un formato numérico adecuado (One-Hot Encoding, Label Encoding, etc.).

- **Discretización de variables**: Conversión de variables numéricas continuas en categorías si son relevantes para el desempeño del modelo a entrenar.

- **Estandarización o normalización** : Ajustar la escala de los datos para mejorar el rendimiento de los algoritmos sensibles a la magnitud de las variables.

- **Eliminación o transformación de valores atípicos**: Identificar y tratar con datos outliers para mejorar la robustez del modelo.

- **Nuevas características** : Creación de variables adicionales que puedan aportar información relevante al modelo.

Cada una de estas transformaciones debe ser justificada en función de su relevancia para el problema y los datos, y es importante evaluar su impacto en el rendimiento del modelo. Además, el pipeline debe ser flexible y modular para poder probar diferentes configuraciones de preprocesamiento.

Inicialmente, se generan las clases necesarias para el pipeline.

Se agrega la clase Básica de IQR

```python
class IQR(BaseEstimator, TransformerMixin):
    def __init__(self, lambd=1.5):
        self.lambd = lambd
    def fit(self, X, y=None):
        return self
    def transform(self, X):
        X = X.copy()
        n_cols = X.select_dtypes(include=[np.number]).columns
        for col in n_cols:
            Q1 = X[col].quantile(0.25); Q3 = X[col].quantile(0.75)
            IQR = Q3 - Q1
            inf = Q1 - self.lambd * IQR; sup = Q3 + self.lambd * IQR
            X.loc[(X[col] < inf) | (X[col] > sup), col] = np.nan
        return X
    def set_output(self, transform='default'):
        return self
```

Luego, tanto items como n_orders, serían una fuente de la variable dependiente si se usan tal cual, y no tiene sentido predecir si vendí en una semana en la que se que vendí 3 productos. Por lo mismo, es necesario mover este tipo de variables una semana en el pasado, para que los datos de la semana 3, en realidad tengan cuanto compró el usuario en la semana 2 y así poder predecir de manera correcta 3. Luego, por la misma razón, es necesario en el pipeline eliminar la variable de items y de n_orders.

```python
class DatosHistoricos(BaseEstimator, TransformerMixin):
    def __init__(self, c="customer_id", p="product_id",
                 week_col="week", items="items", n_orders="n_orders", compra="compra"):
        self.c, self.p, self.week_col = c, p, week_col
        self.items, self.n_orders, self.compra = items, n_orders, compra

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        df = X.copy().sort_values([self.c, self.p, self.week_col]).reset_index(drop=True)

        #Cuantos items del producto y, compro el cliente x en la semana t-1.
        df["items_prev"]    = df.groupby([self.c, self.p], sort=False)[self.items].shift(1).fillna(0)
        #Cuantos ordenes del producto y hizo el cliente x en la semana t-1.
        df["n_orders_prev"] = df.groupby([self.c, self.p], sort=False)[self.n_orders].shift(1).fillna(0)
        #Si el cliente compró o no en la semana anterior.
        df["compra_prev"]   = df.groupby([self.c, self.p], sort=False)[self.compra].shift(1).fillna(0).astype("Int8")

        return df
```

Luego, es importante generar variables de recencia, las cuales son cuanto tiempo ha pasado desde que un cliente no ha hecho una compra. En este caso se define el inicio de la recencia como 1000, y como se vió anteriormente, si un cliente compra, no pasa mucho tiempo en que vuelva a comprar.

```python

class RecenciaSemanal(BaseEstimator, TransformerMixin):
    def __init__(self, c="customer_id", p="product_id", week_col="week", compra="compra", inicio=1000):
        self.c, self.p, self.week_col, self.compra = c, p, week_col, compra
        self.inicio = inicio

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        df = X.copy().sort_values([self.c, self.p, self.week_col])
        last_prod = df.groupby([self.c, self.p], sort=False).apply(
            lambda g: g[self.week_col].where(g[self.compra].eq(1)).shift(1).ffill()
        ).reset_index(level=[0,1], drop=True)
        df["recencia_producto"] = (df[self.week_col] - last_prod).fillna(self.inicio).astype("Int32")
        last_any = df.groupby(self.c, sort=False).apply(
            lambda g: g[self.week_col].where(g[self.compra].eq(1)).shift(1).ffill()
        ).reset_index(level=self.c, drop=True)
        df["recencia_general"] = (df[self.week_col] - last_any).fillna(self.inicio).astype("Int32")
        return df
```

Además, podría ser relevante saber si el producto está siendo popular actualmente entre los clientes, nuevamente es necesario ver lo de la semana anterior, ya que ver la semana actual sería filtrar los datos en el modelo.


```python
class PopularidadProductoPrev(BaseEstimator, TransformerMixin):
    def __init__(self, p="product_id", week_col="week", compra="compra"):
        self.p, self.week_col, self.compra = p, week_col, compra

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        df = X.copy()
        df = df.sort_values([self.p, self.week_col])


        n_comp = df.groupby([self.p, self.week_col], sort=False)[self.compra].sum()


        pop_prev = n_comp.groupby(level=0, sort=False).shift(1).fillna(0).astype("float32").reset_index(name="popularity_prev")


        df = df.merge(pop_prev, on=[self.p, self.week_col], how="left")
        df["popularity_prev"] = df["popularity_prev"].fillna(0).astype("float32")

        return df

```

Esta clase incluye dos variables para calcular el ratio de cada cuanto compra el producto "y", además de calcular de cada cuantas compras de producto, compra "y".

```python
class CompraRelativa(BaseEstimator, TransformerMixin):
    def __init__(self, c="customer_id", p="product_id",
                 week_col="week", items="items", compra="compra"):
        self.c, self.p, self.week_col, self.items, self.compra = c, p, week_col, items, compra

    def fit(self, X, y=None): return self

    def transform(self, X):
        df = X.copy().sort_values([self.c, self.p, self.week_col])
        g_cp, g_c = df.groupby([self.c, self.p], sort=False), df.groupby(self.c, sort=False)

        # Frecuencia relativa del producto según cuantas semanas ha comprado, hasta la semana anterior (t-1)
        df["frec_producto"] = (g_cp[self.compra].shift(1).cumsum() /
                               g_cp.cumcount().replace(0, np.nan)).fillna(0)

        # Ratio del producto en relación a los otros productos, hasta la semana anterior (t-1). Por ejemplo si ha comprado 1 vez A y dos veces B, entonces A tiene un ratio de 0.33.
        df["ratio_compra"] = (g_cp[self.items].shift(1).cumsum() /
                              g_c[self.items].shift(1).cumsum().replace(0, np.nan)).fillna(0)

        return df
```

Finalmente, agregamos variables del tipo Media Movil, con 4 semanas, para saber que tan activo ha estado el cliente en el último mes.

```python
class MediaMovil4Semanas(BaseEstimator, TransformerMixin):
    def __init__(self, c="customer_id", p="product_id", week_col="week", items="items", n_orders="n_orders", ventana=4):
        self.c, self.p, self.week_col = c, p, week_col
        self.items, self.n_orders, self.ventana = items, n_orders, ventana

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        df = X.copy().sort_values([self.c, self.p, self.week_col])

        df["MediaMovil_items"] = df.groupby(self.c, sort=False)[self.items].apply(
            lambda s: s.shift(1).rolling(self.ventana, min_periods=1).mean()
        ).astype("float32").values

        df["MediaMovil_ordenes"] = df.groupby(self.c, sort=False)[self.n_orders].apply(
            lambda s: s.shift(1).rolling(self.ventana, min_periods=1).mean()
        ).astype("float32").values

        df["MediaMovil_items_Prod"] = df.groupby([self.c, self.p], sort=False)[self.items].apply(
            lambda s: s.shift(1).rolling(self.ventana, min_periods=1).mean()
        ).astype("float32").values

        df["MediaMovil_ordenes_Prod"] = df.groupby([self.c, self.p], sort=False)[self.n_orders].apply(
            lambda s: s.shift(1).rolling(self.ventana, min_periods=1).mean()
        ).astype("float32").values

        df[["MediaMovil_items","MediaMovil_ordenes","MediaMovil_items_Prod","MediaMovil_ordenes_Prod"]] = \
            df[["MediaMovil_items","MediaMovil_ordenes","MediaMovil_items_Prod","MediaMovil_ordenes_Prod"]].fillna(0.0)

        return df

```
